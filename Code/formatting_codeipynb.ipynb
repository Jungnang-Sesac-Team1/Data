```python
import pandas as pd

def create_rag_document(row):
    # NaN(비어있는 값) 처리: 비어있으면 빈 문자열('')이나 '-'로 대체하여 에러 방지
    title = str(row['공고제목']) if pd.notna(row['공고제목']) else "제목 없음"
    region = str(row['지역']) if pd.notna(row['지역']) else "지역 무관"
    target = str(row['지원대상']) if pd.notna(row['지원대상']) else "전체"
    category = str(row['카테고리']) if pd.notna(row['카테고리']) else "일반"
    content = str(row['공고내용']) if pd.notna(row['공고내용']) else ""
    content = content.replace('\ufffd', '')
    content = content.replace("\n", " ").replace("  ", " ").strip()
    method = str(row['사업신청방법설명']) if pd.notna(row['사업신청방법설명']) else "공고문 참조"
    contact = str(row['문의처']) if pd.notna(row['문의처']) else "문의처 확인 필요"
    period = str(row['접수기간']) if pd.notna(row['접수기간']) else "별도 안내"
    period_suffix = "입니다." if str(period).strip().endswith("까지") else "까지입니다."

    # 1. 텍스트 생성 (검색용)
    # yugwan_df의 컬럼들을 문장으로 자연스럽게 연결
    text = (
        f"이 공고의 제목은 '{title}'이며, 분류는 '{category}'에 해당합니다. "
        f"주로 {region} 지역의 {target}을(를) 대상으로 지원합니다. "
        f"주요 공고 내용은 {content} 등이며, 접수 기간은 {period}{period_suffix} "
        f"신청 방법은 {method}을(를) 따르며, 자세한 문의처는 {contact}입니다."
    )
    
    # 2. 메타데이터 생성 (필터링용)
    # 지역, 카테고리 등을 나중에 필터 옵션으로 쓰기 위해 분리 저장
    metadata = {
        "title": title,
        "region": region,         
        "category": category,      
        "target": target,         
        "deadline": period         
    }
    
    return text, metadata

# -----------------------------------------------------------
# [실행 및 확인] yugwan_df의 첫 번째 행으로 테스트
# -----------------------------------------------------------

if not yugwan_df.empty:
    # 1. 첫 번째 행 가져오기
    first_row = yugwan_df.iloc[0]
    
    # 2. 함수 실행
    rag_text, rag_metadata = create_rag_document(first_row)

    print("=== [RAG에 저장될 텍스트 (Text Chunk)] ===")
    print(rag_text)
    print("\n=== [함께 저장될 메타데이터 (Metadata)] ===")
    print(rag_metadata)
    
    # (참고) 전체 데이터프레임에 적용하여 리스트로 만들고 싶다면:
    # results = yugwan_df.apply(lambda x: create_rag_document(x), axis=1)
    # texts = [r[0] for r in results]
    # metadatas = [r[1] for r in results]
```

```python
import pandas as pd

# -----------------------------------------------------------
# 1. 포맷팅 함수 정의
# -----------------------------------------------------------
def create_loan_rag_document(row):
    """
    금융 상품 데이터를 LLM이 이해하기 쉬운 자연어 문장으로 변환하고,
    필터링을 위한 메타데이터를 추출합니다.
    """
    
    # [A] 데이터 안전하게 가져오기 (결측치 및 '-' 처리 헬퍼 함수)
    def get_val(col, default="정보 없음"):
        # 컬럼이 없거나 값이 비어있는 경우 처리
        if col not in row.index:
            return default
        val = row[col]
        if pd.isna(val) or str(val).strip() in ['-', '', 'nan']:
            return default
        return str(val).strip()

    # [B] 주요 필드 추출
    title = get_val('공고제목', '제목 미정')
    region = get_val('지역', '전국')
    category = get_val('카테고리', '금융지원')
    
    # 기관명 처리 (제공기관/취급기관이 같이 있는 경우 앞부분만 사용 등)
    inst_raw = get_val('취급기관')
    prov_raw = get_val('제공기관/취급기관')
    institution = f"{inst_raw} / {prov_raw.split('/')[0].strip()}" if prov_raw != "정보 없음" else inst_raw
    
    # 금리 및 한도 처리
    limit = get_val('최대한도(만원)')
    if limit != "정보 없음" and str(limit).isdigit(): limit = f"{limit}만원"
    
    rate = get_val('금리(%)')
    rate_desc = f"{rate}%" if rate != "정보 없음" else "금리는 별도 문의가 필요합니다"

    # 기간 및 상환
    period = get_val('대출기간(년)')
    if str(period).isdigit(): period = f"{period}년"
    repay_method = get_val('상환방법')
    
    # [C] 핵심!! 자격 요건 상세화
    target_main = get_val('지원대상')
    condition_detail = get_val('지원대상 상세조건', '')
    income_cond = get_val('소득기준', '')
    credit_cond = get_val('신용조건', '')
    
    # 조건들을 자연스러운 문장으로 연결
    eligibility_text = f"기본 지원 대상은 {target_main}입니다."
    if condition_detail != "정보 없음":
        eligibility_text += f" 상세 자격 요건은 다음과 같습니다. {condition_detail}."
    if income_cond != "정보 없음" and income_cond not in condition_detail:
        eligibility_text += f" 소득 기준으로는 {income_cond}이어야 합니다."
    if credit_cond != "정보 없음" and credit_cond not in condition_detail:
        eligibility_text += f" 신용 조건은 {credit_cond}입니다."

    # 비용 및 기타
    costs = get_val('대출부대비용')
    fees = get_val('중도상환수수료')
    etc_notes = get_val('기타 참고사항')
    
    contact = get_val('문의처')
    deadline = get_val('운영기한')
    url = get_val('관련사이트', '')

    # -----------------------------------------------------------
    # [D] 텍스트 청크 생성 (RAG 검색용 상세 텍스트)
    # -----------------------------------------------------------
    rag_text = (
        f"이 상품은 {region} 지역의 '{title}' 공고입니다. "
        f"카테고리는 {category}이며, 주요 취급 및 제공 기관은 {institution}입니다.\n\n"
        
        f"대출 최대 한도는 {limit}이며, {rate_desc}. "
        f"대출 기간은 {period}이며, 상환 방식은 {repay_method}입니다. "
        f"(총 대출 기간 참고: {get_val('총대출기간(년)')})\n\n"
        
        f"{eligibility_text}\n\n"
        
        f"대출 부대 비용으로는 {costs}이 발생하며, 중도상환수수료는 {fees}입니다. "
        f"기타 참고사항으로 {etc_notes} 조건이 있습니다.\n\n"
        
        f"신청 및 문의는 {contact}로 가능하며, 운영 기한은 {deadline}입니다."
    )

    # -----------------------------------------------------------
    # [E] 메타데이터 생성 (필터링용)
    # -----------------------------------------------------------
    metadata = {
        "title": title,                 # 제목
        "region": region,               # 지역
        "category": category,           # 카테고리
        "target": target_main,          # 지원대상
        "deadline": deadline,             # 마감기한
    }

    return rag_text, metadata

# -----------------------------------------------------------
# [실행 및 확인] loan_df의 첫 번째 행(iloc[0])으로 테스트
# -----------------------------------------------------------

# loan_df가 존재하는지 확인 후 실행
if 'loan_df' in locals() and not loan_df.empty:
    print(f"✅ loan_df가 로드되었습니다. 총 {len(loan_df)}개의 데이터가 있습니다.")
    print("▶ 첫 번째 행 데이터를 사용하여 포맷팅을 테스트합니다.\n")
    
    # 1. 첫 번째 행 가져오기
    first_row = loan_df.iloc[0]

    # 2. 함수 적용
    text_result, meta_result = create_loan_rag_document(first_row)

    # 3. 결과 출력
    print("====== [Generated RAG Text (Text Chunk)] ======")
    print(text_result)
    print("\n====== [Generated Metadata] ======")
    print(meta_result)

else:
    print("⚠️ 'loan_df' 변수를 찾을 수 없습니다. CSV 파일을 먼저 로드해주세요.")
    print("예: loan_df = pd.read_csv('loan_final.csv', encoding='cp949')")
```

```python
import pandas as pd
import json

# -----------------------------------------------------------
# 1. 변환 함수 정의
# -----------------------------------------------------------
def create_rag_json_item(row):
    # [A] 데이터 안전하게 가져오기 (NaN 처리)
    def get_val(col, default=""):
        if col not in row.index: return default
        val = row[col]
        return str(val).strip() if pd.notna(val) else default

    # [B] 필드 추출
    title = get_val('공고제목', '제목 없음')
    region = get_val('지역', '지역 무관')
    target = get_val('지원대상', '전체')
    category = get_val('카테고리', '일반')
    
    # 내용 전처리
    content_raw = get_val('공고내용')
    content = content_raw.replace('\ufffd', '').replace("\n", " ").replace("  ", " ").strip()
    
    method = get_val('사업신청방법설명', '공고문 참조')
    contact = get_val('문의처', '문의처 확인 필요')
    period = get_val('접수기간', '별도 안내')
    
    # 문장 끝맺음 처리
    period_suffix = "입니다." if period.endswith("까지") else "까지입니다."

    # [C] content 텍스트 생성
    text_content = (
        f"이 공고의 제목은 '{title}'이며, 분류는 '{category}'에 해당합니다. "
        f"주로 {region} 지역의 {target}을(를) 대상으로 지원합니다. "
        f"주요 공고 내용은 {content} 등이며, 접수 기간은 {period}{period_suffix} "
        f"신청 방법은 {method}을(를) 따르며, 자세한 문의처는 {contact}입니다."
    )
    
    # [D] metadata 생성
    metadata_dict = {
        "title": title,
        "region": region,
        "category": category,
        "target": target,
        "deadline": period
    }
    
    # [E] 최종 구조 반환
    return {
        "metadata": metadata_dict,
        "content": text_content
    }

# -----------------------------------------------------------
# 2. 실행 및 JSON 파일 내보내기
# -----------------------------------------------------------

# yugwan_df가 존재하는지 체크
if 'yugwan_df' in locals() and not yugwan_df.empty:
    
    # 1. 데이터 변환 (DataFrame -> List)
    rag_json_list = yugwan_df.apply(create_rag_json_item, axis=1).tolist()
    print(f"✅ 변환 완료! 총 {len(rag_json_list)}개의 데이터가 준비되었습니다.")

    # 2. JSON 파일로 저장 (내보내기)
    output_filename = "yugwan_rag_data.json"
    
    try:
        with open(output_filename, "w", encoding="utf-8") as f:
            # ensure_ascii=False: 한글이 깨지지 않게 저장
            # indent=4: 사람이 보기 좋게 들여쓰기 적용
            json.dump(rag_json_list, f, ensure_ascii=False, indent=4)
            
        print(f"💾 파일 저장 성공: {output_filename}")
        
    except Exception as e:
        print(f"❌ 파일 저장 실패: {e}")

else:
    print("⚠️ 'yugwan_df' 변수가 메모리에 없습니다. 데이터프레임을 먼저 로드해주세요.")
```

```python
import pandas as pd
import json

# -----------------------------------------------------------
# 1. 사용자 정의 함수 (작성하신 코드 원본 유지)
# -----------------------------------------------------------
def create_sojingong_rag_document(row):
    # 1. 데이터 안전하게 가져오기 (NaN 처리)
    title = str(row['공고제목']) if pd.notna(row['공고제목']) else "제목 없음"
    category = str(row['카테고리']) if pd.notna(row['카테고리']) else "소진공 공고"
    region = str(row['지역']) if pd.notna(row['지역']) else "전국/무관"
    target = str(row['지원대상']) if pd.notna(row['지원대상']) else "소상공인"
    content = str(row['공고내용']) if pd.notna(row['공고내용']) else "내용 없음"
    content = content.replace('\ufffd', '')
    content = content.replace("\n", " ").replace("  ", " ").strip()
    period = str(row['접수기간']) if pd.notna(row['접수기간']) else "별도 안내"

    # 2. '까지' 중복 방지 로직 (요청하신 부분)
    period_suffix = "입니다." if str(period).strip().endswith("까지") else "까지입니다."

    # 3. 텍스트 생성 (검색용)
    # 소진공 데이터는 '공고내용'에 핵심 정보가 많으므로 이를 강조합니다.
    text = (
        f"이 공고의 제목은 '{title}'이며, '{category}' 카테고리에 해당합니다. "
        f"지원 지역은 {region}이며, 지원 대상은 {target}입니다. "
        f"접수 기간은 {period}{period_suffix} "
        f"주요 공고 내용은 다음과 같습니다: {content}"
    )
    
    # 4. 메타데이터 생성 (필터링용)
    metadata = {
        "title": title,
        "region": region,
        "category": category,
        "target": target,
        "deadline": period
    }
    
    return text, metadata

# -----------------------------------------------------------
# 2. JSON 구조 변환 래퍼(Wrapper) 함수
# -----------------------------------------------------------
def convert_to_rag_format(row):
    # 작성하신 함수의 결과(Tuple)를 받아옵니다.
    text_result, meta_result = create_sojingong_rag_document(row)
    
    # 요청하신 Dictionary 형태로 재구성하여 반환합니다.
    return {
        "metadata": meta_result,
        "content": text_result
    }

# -----------------------------------------------------------
# 3. 실행 및 JSON 파일 내보내기
# -----------------------------------------------------------

# sojingong_df가 존재하는지 체크
if 'sojingong_df' in locals() and not sojingong_df.empty:
    
    # 1. 데이터 변환 (DataFrame -> List of Dicts)
    # apply를 사용하여 각 행마다 포맷팅을 수행하고 리스트로 변환
    rag_json_list = sojingong_df.apply(convert_to_rag_format, axis=1).tolist()
    
    print(f"✅ 변환 완료! 총 {len(rag_json_list)}개의 데이터가 준비되었습니다.")

    # 2. JSON 파일로 저장 (내보내기)
    output_filename = "sojingong_rag_data.json"
    
    try:
        with open(output_filename, "w", encoding="utf-8") as f:
            # ensure_ascii=False: 한글 깨짐 방지
            # indent=4: 보기 좋게 들여쓰기
            json.dump(rag_json_list, f, ensure_ascii=False, indent=4)
            
        print(f"💾 파일 저장 성공: {output_filename}")
        
    except Exception as e:
        print(f"❌ 파일 저장 실패: {e}")

else:
    print("⚠️ 'sojingong_df' 변수가 메모리에 없습니다. 데이터프레임을 먼저 로드해주세요.")
```

```python
import pandas as pd
import json

# -----------------------------------------------------------
# 1. 변환 함수 정의 (Row -> Dict {metadata, content})
# -----------------------------------------------------------
def create_loan_rag_json_item(row):
    # [A] 데이터 안전하게 가져오기 (결측치 및 '-' 처리 헬퍼 함수)
    def get_val(col, default="정보 없음"):
        # 컬럼이 없거나 값이 비어있는 경우 처리
        if col not in row.index:
            return default
        val = row[col]
        if pd.isna(val) or str(val).strip() in ['-', '', 'nan']:
            return default
        return str(val).strip()

    # [B] 주요 필드 추출
    title = get_val('공고제목', '제목 미정')
    region = get_val('지역', '전국')
    category = get_val('카테고리', '금융지원')
    
    # 기관명 처리
    inst_raw = get_val('취급기관')
    prov_raw = get_val('제공기관/취급기관')
    institution = f"{inst_raw} / {prov_raw.split('/')[0].strip()}" if prov_raw != "정보 없음" else inst_raw
    
    # 금리 및 한도 처리
    limit = get_val('최대한도(만원)')
    if limit != "정보 없음" and str(limit).isdigit(): limit = f"{limit}만원"
    
    rate = get_val('금리(%)')
    rate_desc = f"{rate}%" if rate != "정보 없음" else "금리는 별도 문의가 필요합니다"

    # 기간 및 상환
    period = get_val('대출기간(년)')
    if str(period).isdigit(): period = f"{period}년"
    repay_method = get_val('상환방법')
    
    # [C] 자격 요건 상세화
    target_main = get_val('지원대상')
    condition_detail = get_val('지원대상 상세조건', '')
    income_cond = get_val('소득기준', '')
    credit_cond = get_val('신용조건', '')
    
    # 조건들을 자연스러운 문장으로 연결
    eligibility_text = f"기본 지원 대상은 {target_main}입니다."
    if condition_detail != "정보 없음":
        eligibility_text += f" 상세 자격 요건은 다음과 같습니다. {condition_detail}."
    if income_cond != "정보 없음" and income_cond not in condition_detail:
        eligibility_text += f" 소득 기준으로는 {income_cond}이어야 합니다."
    if credit_cond != "정보 없음" and credit_cond not in condition_detail:
        eligibility_text += f" 신용 조건은 {credit_cond}입니다."

    # 비용 및 기타
    costs = get_val('대출부대비용')
    fees = get_val('중도상환수수료')
    etc_notes = get_val('기타 참고사항')
    
    contact = get_val('문의처')
    deadline = get_val('운영기한')
    url = get_val('관련사이트', '')

    # [D] 텍스트 청크 생성 (content)
    rag_text = (
        f"이 상품은 {region} 지역의 '{title}' 공고입니다. "
        f"카테고리는 {category}이며, 주요 취급 및 제공 기관은 {institution}입니다.\n\n"
        
        f"대출 최대 한도는 {limit}이며, {rate_desc}. "
        f"대출 기간은 {period}이며, 상환 방식은 {repay_method}입니다. "
        f"(총 대출 기간 참고: {get_val('총대출기간(년)')})\n\n"
        
        f"{eligibility_text}\n\n"
        
        f"대출 부대 비용으로는 {costs}이 발생하며, 중도상환수수료는 {fees}입니다. "
        f"기타 참고사항으로 {etc_notes} 조건이 있습니다.\n\n"
        
        f"신청 및 문의는 {contact}로 가능하며, 운영 기한은 {deadline}입니다."
    )

    # [E] 메타데이터 생성 (metadata)
    metadata = {
        "title": title,
        "region": region,
        "category": category,
        "target": target_main,
        "deadline": deadline,
    }

    # [F] 최종 구조 반환
    return {
        "metadata": metadata,
        "content": rag_text
    }

# -----------------------------------------------------------
# 2. 실행 및 JSON 파일 내보내기
# -----------------------------------------------------------

# loan_df가 존재하는지 체크
if 'loan_df' in locals() and not loan_df.empty:
    
    # 1. 데이터 변환 (DataFrame -> List of Dicts)
    # apply를 사용하여 각 행마다 포맷팅을 수행하고 리스트로 변환
    rag_json_list = loan_df.apply(create_loan_rag_json_item, axis=1).tolist()
    
    print(f"✅ 변환 완료! 총 {len(rag_json_list)}개의 데이터가 준비되었습니다.")

    # 2. JSON 파일로 저장 (내보내기)
    output_filename = "loan_rag_data.json"
    
    try:
        with open(output_filename, "w", encoding="utf-8") as f:
            # ensure_ascii=False: 한글 깨짐 방지
            # indent=4: 보기 좋게 들여쓰기
            json.dump(rag_json_list, f, ensure_ascii=False, indent=4)
            
        print(f"💾 파일 저장 성공: {output_filename}")
        
    except Exception as e:
        print(f"❌ 파일 저장 실패: {e}")

else:
    print("⚠️ 'loan_df' 변수가 메모리에 없습니다. 데이터프레임을 먼저 로드해주세요.")
```

```python
import json
import os

# 1. 합칠 파일 목록 정의
target_files = [
    'loan_rag_data.json',
    'yugwan_rag_data.json',
    'sojingong_rag_data.json'
]

merged_data = []

print("=== 파일 병합 시작 ===")

# 2. 각 파일을 순회하며 데이터 읽기 및 병합
for file_name in target_files:
    if os.path.exists(file_name):
        try:
            with open(file_name, 'r', encoding='utf-8') as f:
                data = json.load(f)
                
                # 데이터가 리스트 형태인지 확인 후 병합
                if isinstance(data, list):
                    count = len(data)
                    merged_data.extend(data) # 리스트 이어 붙이기
                    print(f"✅ {file_name} 로드 성공: {count}개 데이터 병합")
                else:
                    print(f"⚠️ {file_name}: 데이터가 리스트 형식이 아닙니다. (병합 건너뜀)")
                    
        except Exception as e:
            print(f"❌ {file_name} 읽기 중 에러 발생: {e}")
    else:
        print(f"⚠️ {file_name} 파일을 찾을 수 없습니다. (이전 단계에서 생성되었는지 확인 필요)")

# 3. 통합된 데이터를 하나의 JSON 파일로 저장
output_filename = "rag_data.json"

try:
    with open(output_filename, 'w', encoding='utf-8') as f:
        json.dump(merged_data, f, ensure_ascii=False, indent=4)
    
    print("-" * 30)
    print(f"🎉 병합 완료! 총 {len(merged_data)}개의 데이터가 '{output_filename}'에 저장되었습니다.")

except Exception as e:
    print(f"❌ 최종 파일 저장 실패: {e}")
```
