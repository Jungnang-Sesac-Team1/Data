```python
# 1. 데이터를 담을 딕셔너리 생성
dfs = {}

# 2. 각 카테고리명 리스트
categories = ['유관사업 공고조회', '소진공 공고조회', '대출상품 조회']

# 3. 반복문을 통해 필터링 및 할당
for cat in categories:
    # 해당 카테고리에 맞는 행만 추출하여 복사본 저장
    dfs[cat] = df_filtered[df_filtered['카테고리'] == cat].copy()
    
    # 인덱스 재설정 (0부터 다시 시작하도록)
    dfs[cat].reset_index(drop=True, inplace=True)
    
    print(f"✅ '{cat}' 데이터프레임 생성 완료: {len(dfs[cat])}건")

# ==========================================
# 4. 개별 변수로 사용하고 싶을 경우
# ==========================================
df_yugwan = dfs['유관사업 공고조회']
df_sojingong = dfs['소진공 공고조회']
df_loan = dfs['대출상품 조회']
```

```python
import pandas as pd
import re

def parse_yugwan_text(text):
    """
    본문 텍스트를 분석하여 항목별로 내용을 추출하는 함수
    """
    # 1. 불필요한 뒷부분 제거 ('카카오톡으로 공유하기' 이후 삭제)
    if "카카오톡으로 공유하기" in text:
        text = text.split("카카오톡으로 공유하기")[0]
    
    # 2. 추출할 키워드 리스트 (순서 중요!)
    # 정규표현식에서 괄호()는 특수문자이므로 앞에 역슬래시(\)를 붙여야 합니다.
    keywords = [
        "공고명", "지원대상", "소관기관", "지원분야\(대\)", "지원분야\(중\)",
        "사업수행기관", "문의처", "신청기간", "공고내용", 
        "사업신청방법설명", "첨부파일"
    ]
    
    parsed_data = {}
    
    for i in range(len(keywords)):
        current_keyword = keywords[i]
        
        # 다음 키워드가 있는지 확인 (마지막 키워드인 '첨부파일' 제외)
        next_keyword = keywords[i+1] if i < len(keywords) - 1 else None
        
        # 딕셔너리에 저장할 깔끔한 키 이름 (괄호 이스케이프 제거)
        clean_key = current_keyword.replace(r"\(", "(").replace(r"\)", ")")
        
        try:
            # 현재 키워드의 위치 찾기
            start_match = re.search(current_keyword, text)
            if not start_match:
                parsed_data[clean_key] = "" # 키워드가 없으면 빈칸
                continue
                
            start_index = start_match.end() # 키워드 끝나는 지점부터 내용 시작
            
            # 다음 키워드 위치 찾기 (내용의 끝 지점)
            end_index = len(text) # 기본값: 텍스트 끝까지
            if next_keyword:
                next_match = re.search(next_keyword, text)
                if next_match:
                    end_index = next_match.start() # 다음 키워드 시작 지점 전까지
            
            # 내용 추출 및 공백 제거
            content = text[start_index:end_index].strip()
            parsed_data[clean_key] = content
            
        except Exception as e:
            parsed_data[clean_key] = ""

    return pd.Series(parsed_data)

# ==========================================
# 실행 코드
# ==========================================

# 1. 빈 데이터프레임 방지 (데이터가 있을 때만 실행)
if not df_yugwan.empty:
    print("⏳ 본문 내용 분리 작업을 시작합니다...")

    # 2. 본문 컬럼에 함수 적용하여 새로운 컬럼들 생성
    # apply를 쓰면 각 행마다 함수가 실행되고, 결과가 새로운 데이터프레임으로 반환됩니다.
    extracted_cols = df_yugwan['본문'].apply(parse_yugwan_text)
    
    # 3. 기존 df_yugwan과 합치기 (기존 '본문' 컬럼은 제거하거나 유지 선택 가능)
    # 여기서는 보기 좋게 '본문'을 빼고 합칩니다.
    df_yugwan_final = pd.concat([df_yugwan.drop('본문', axis=1), extracted_cols], axis=1)
    
    print("✅ 분리 완료!")
    
    # 4. 결과 확인
    print("\n--- [분리된 데이터 미리보기] ---")
    # 화면에 잘 보이게 공고명과 지원대상만 먼저 출력
    print(df_yugwan_final[['공고제목', '지원대상', '소관기관', '첨부파일']].head(3))
    
    # 전체 컬럼명 확인
    print("\n--- [생성된 컬럼 목록] ---")
    print(df_yugwan_final.columns.tolist())

else:
    print("⚠️ df_yugwan 데이터가 비어있습니다.")
```

```python
import pandas as pd
import re

def parse_sojingong_text(text):
    """
    소진공 공고조회 본문 텍스트를 분석하여 항목별로 추출하는 함수
    """
    # 1. 불필요한 뒷부분 제거 ('카카오톡으로 공유하기' 이후 삭제)
    if "카카오톡으로 공유하기" in text:
        text = text.split("카카오톡으로 공유하기")[0]
    
    # 2. 텍스트 내 실제 등장하는 키워드 순서 (그대로 적어야 함)
    # 주의: 텍스트에는 '모집유형'으로 적혀있지만, 나중에 '공고대상'으로 저장할 것입니다.
    # 주의: '공고문 미리보기'는 띄어쓰기가 포함되어 있습니다.
    raw_keywords = [
        "공고명", "모집유형", "지원사업유형명", "사업기간", "접수기간",
        "공고내용", "공고문 미리보기", "첨부파일", "연관주제어", "상담신청기간"
    ]
    
    # 3. (옵션) 텍스트의 키워드 -> 내가 원하는 컬럼명으로 변경 매핑
    col_mapping = {
        "모집유형": "공고대상",          # 요청하신 이름으로 변경
        "공고문 미리보기": "공고문미리보기" # 띄어쓰기 제거
    }
    
    parsed_data = {}
    
    for i in range(len(raw_keywords)):
        current_keyword = raw_keywords[i]
        
        # 다음 키워드가 있는지 확인
        next_keyword = raw_keywords[i+1] if i < len(raw_keywords) - 1 else None
        
        # 저장할 컬럼명 결정 (매핑에 있으면 그거 쓰고, 없으면 원래 이름 사용)
        save_col_name = col_mapping.get(current_keyword, current_keyword)
        
        try:
            # 현재 키워드 위치 찾기
            start_match = re.search(current_keyword, text)
            if not start_match:
                parsed_data[save_col_name] = ""
                continue
                
            start_index = start_match.end() # 키워드 끝나는 지점
            
            # 다음 키워드 위치 찾기
            end_index = len(text)
            if next_keyword:
                next_match = re.search(next_keyword, text)
                if next_match:
                    end_index = next_match.start()
            
            # 내용 추출 및 공백 제거
            content = text[start_index:end_index].strip()
            parsed_data[save_col_name] = content
            
        except Exception as e:
            parsed_data[save_col_name] = ""

    return pd.Series(parsed_data)

# ==========================================
# 실행 코드
# ==========================================

if not df_sojingong.empty:
    print("⏳ 소진공 데이터 분리 작업을 시작합니다...")

    # 1. 함수 적용
    extracted_cols = df_sojingong['본문'].apply(parse_sojingong_text)
    
    # 2. 기존 df_sojingong과 합치기 (기존 '본문' 제거)
    df_sojingong_final = pd.concat([df_sojingong.drop('본문', axis=1), extracted_cols], axis=1)
    
    print("✅ 소진공 데이터 정제 완료!")
    
    # 3. 결과 확인
    # 요청하신 주요 컬럼 미리보기
    target_cols_preview = ['공고명', '공고대상', '접수기간', '첨부파일']
    print("\n--- [분리된 데이터 미리보기] ---")
    print(df_sojingong_final[target_cols_preview].head(3))
    
    # 전체 컬럼명 확인
    print("\n--- [생성된 컬럼 목록] ---")
    print(df_sojingong_final.columns.tolist())

else:
    print("⚠️ df_sojingong 데이터가 비어있습니다.")
```
