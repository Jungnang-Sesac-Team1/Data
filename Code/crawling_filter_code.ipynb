```python
import pandas as pd
import os

file_path = "merged_crawling_result.csv"

if os.path.exists(file_path):
    # 1. 파일 불러오기
    df = pd.read_csv(file_path, encoding="utf-8-sig")
    
    # 2. NaN 값 처리 (에러 방지용)
    df = df.fillna("")

    # ======================================================
    # [Step 1] 컬럼명(Index) 변경
    # ======================================================
    # 현재 순서: [제목, 검색어, URL, 본문, 첨부파일]
    # 변경 순서: [카테고리, 공고제목, URL, 본문, 첨부파일]
    df.columns = ["카테고리", "공고제목", "URL", "본문", "첨부파일"]
    
    # ======================================================
    # [Step 2] 본문 내 줄바꿈(\n) 제거 
    # ======================================================
    df['본문'] = df['본문'].str.replace("\n", " ", regex=False)

    print("✅ 데이터 정제 완료!")
    
    # 3. 변경된 결과 확인
    print("\n--- [수정된 데이터 미리보기] ---")
    print(df[['카테고리', '공고제목']].head()) # 잘 바뀌었는지 확인

else:
    print("⚠️ 파일이 없습니다.")
```

```python
df
```

```python
keywords = "소상공인|소공인"
mask = df['공고제목'].str.contains(keywords, na=False) | df['카테고리'].str.contains(keywords, na=False)
df_filtered = df[mask].copy() # .copy()는 경고 방지용 안전 장치

    # ======================================================
    # [추가됨] 인덱스 재설정 (0부터 순서대로)
    # ======================================================
    # drop=True: 기존의 뒤죽박죽된 인덱스를 별도 컬럼으로 남기지 않고 삭제함
df_filtered.reset_index(drop=True, inplace=True)

print(f"✅ 필터링 후 데이터 개수: {len(df_filtered)}개")
    
    # 3. 결과 확인
print("\n--- [필터링 및 인덱스 재설정 결과] ---")
print(df_filtered.head())
    
    # (확인용) 인덱스가 0, 1, 2... 로 잘 바뀌었는지 출력
    # print(df_filtered.index)
```
